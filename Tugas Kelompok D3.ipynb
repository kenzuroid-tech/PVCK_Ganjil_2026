```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

path_ktm_wajah = ("/content/drive/MyDrive/SEMESTER 5/Citra/KTM (Kartu Tanda Mahasiswa).jpg")
img_bgr = cv2.imread(path_ktm_wajah)

# Pengecekan keamanan
if img_bgr is None:
  print("Gambar tidak ditemukan! Pastikan path file benar atau file sudah di-upload.")
else:
  # Konversi untuk proses deteksi (Gray) dan visualisasi (RGB)
  img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
  img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

  #Load Model Haar Cascade Deteksi Wajah Bawaan OpenCV
  face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")

  #Deteksi Lokasi Wajah
  faces = face_cascade.detectMultiScale(img_gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

  #Proses Menutupi (Sensor) Bagian Wajah
  img_sensored = img_rgb.copy()

  # Jika wajah ditemukan
  if len(faces) > 0:
    for x, y, w, h in faces:
      img_sensored[y : y + h, x : x + w] = [0, 0, 0]

    print(f"Berhasil mendeteksi dan menutupi {len(faces)} area wajah.")
  else:
    print("Wajah tidak terdeteksi pada gambar ini.")

  # 5. Tampilkan Hasil
  plt.figure(figsize=(10, 6))

  plt.subplot(1, 2, 1)
  plt.imshow(img_rgb)
  plt.title("Gambar KTM Asli")
  plt.axis("off")

  plt.subplot(1, 2, 2)
  plt.imshow(img_sensored)
  plt.title("KTM dengan Wajah Disensor (Hitam)")
  plt.axis("off")

  plt.tight_layout()
  plt.show()
```
