```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,0/0/default.jpg"]
# Baca dan tampilkan image
# Loop dari tiap url image, beberapa image dapat disimpan pada list
for url in urls:
  image = io.imread(url)                          # read image
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5) # resize image to half size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)  # convert image to RGB
  final_frame = cv.hconcat((image, image_2))      # concaenate image
  cv2_imshow(final_frame)
  print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image : tinggi x lebar = ",tinggi, "x", "lebar")
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# membuat garis horizontal ditengah image
for y in range(lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# MEMBUAT GARIS VERTIKAL DI TENGAH IMAGE
# Perulangan berjalan sepanjang tinggi gambar (y dari atas ke bawah)
for y in range(tinggi):
  # Koordinat OpenCV diakses dengan format: [baris, kolom] atau [y, x]
  image_3[y, int((lebar)/2)] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Tentukan koordinat 4 pojok gambar (Format OpenCV: X, Y)
pojok_kiri_atas  = (0, 0)
pojok_kanan_bawah = (lebar - 1, tinggi - 1)

pojok_kiri_bawah  = (0, tinggi - 1)
pojok_kanan_atas = (lebar - 1, 0)

warna = (255, 255, 255) # Putih
ketebalan = 1

# Menggambar dua garis diagonal yang saling bersilangan
cv.line(image_3, pojok_kiri_atas, pojok_kanan_bawah, warna, ketebalan)
cv.line(image_3, pojok_kiri_bawah, pojok_kanan_atas, warna, ketebalan)

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# membuat garis horizontal ditengah image
for y in range(280, 400):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# 1. Tentukan koordinat pojok kiri atas (x1, y1) dan kanan bawah (x2, y2)
# Format koordinat OpenCV: (X, Y) atau (Kolom, Baris)
pojok_kiri_atas  = (100, 100)  # Mulai dari x=100, y=100
pojok_kanan_bawah = (400, 300) # Selesai di x=400, y=300

warna = (255, 255, 255) # Putih
ketebalan = 3           # Ketebalan garis kotak dalam piksel

cv.rectangle(image_3, pojok_kiri_atas, pojok_kanan_bawah, warna, -1)

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```
